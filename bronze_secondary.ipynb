```python
from pyspark.sql import functions as F

vol   = "/Volumes/capstone6_telecom/telecom/ext_raw_landing"
src   = f"{vol}/adls/ secondary_events/"  # source folder name has a leading space
table = "capstone6_telecom.telecom.bronze_secondary_events"

df = (spark.readStream.format("cloudFiles")
      .option("cloudFiles.format", "csv")
      .option("header", "true")
      .option("cloudFiles.schemaLocation", f"{vol}/_schemas/secondary_events")
      .option("cloudFiles.schemaEvolutionMode", "addNewColumns")
      .option("cloudFiles.inferColumnTypes", "true")
      .option("cloudFiles.schemaHints", "event_value DOUBLE")
      .load(src)
      .withColumn("source_file", F.col("_metadata.file_path"))
      .withColumn("ingest_ts", F.current_timestamp()))

(df.writeStream
   .option("checkpointLocation", f"{vol}/_checkpoints/secondary_events")
   .option("mergeSchema", "true")
   .trigger(availableNow=True)
   .toTable(table)
   .awaitTermination())
```
